sparse_matrix only stores the non zero values of a matrix. x is an identity matrix and it is converted to sparse matrix and only the 1's are stored with their index.
Now converrsely using the same idea, a matrix can be created where we store some values from a dataset(array) into specific positions of a matrix and rest of the positions will automatically be filled to be zero

```python
data = [1,2,3,4]
row = [0,1,0,2]   #row positions
column = [0,2,2,1]  #column postions
csr = sparse.csr_matrix((data , (row,column)) , shape = (3,3))
print(csr)
```

```python
#convert to csr to array
print(csr.toarray())
```
