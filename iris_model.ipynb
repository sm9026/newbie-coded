```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

**training data visualization using pair plot**

```python
#create data frame from data in X_train
#Label the colums using strings in iris_dataset.feature_names
iris_dataframe = pd.DataFrame(X_train , columns = iris_dataset.feature_names)

#create a scatter-matrix (pair plot) from the dataframe , the colour of the points is referred from y_train
scatter = pd.plotting.scatter_matrix(
                           iris_dataframe , 
                           c = y_train , 
                           figsize=(12,12) , 
                           diagonal = 'hist' , 
                           hist_kwds = {'bins':20 , 'edgecolor' : 'black' , 'alpha' : 0.7} ,
                           s=60 ,
                           marker = 'o' ,
                           cmap = 'Set1'
)
for ax in scatter.ravel():
    ax.grid(False)
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.suptitle('Iris Dataset - Training Set Scatter Matrix' , fontsize=16, fontweight='bold', y=0.995)

plt.tight_layout()
plt.show()
```
